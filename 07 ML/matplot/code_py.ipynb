```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
marks = np.array([12, 15, 18, 22, 25, 27, 33, 35, 38, 41, 45, 52, 57, 63, 68, 74, 81, 88, 92])
plt.title("Histogram")
plt.hist(marks, bins=5, edgecolor="black")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```
