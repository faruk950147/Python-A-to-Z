```python
import matplotlib.pyplot as plt
import numpy as np  

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 3, 5, 7, 11])
# plt.plot(x, y) thats means x of 1 will be connected with y of 2 → (1,2) 
# x of 1 of y of 2 connected → (1,2)
# x of 2 of y of 3 connected → (2,3)
# x of 3 of y of 5 connected → (3,5)
# x of 4 of y of 7 connected → (4,7)
# x of 5 of y of 11 connected → (5,11)
plt.plot(x, y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Line Plot')
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np  
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 3, 5, 7, 11])
# plt.scatter(x, y) thats means x of 1 will be connected with y of 2 → (1,2)
# x of 1 of y of 2 connected → (1,2)
# x of 2 of y of 3 connected → (2,3)
#  x of 3 of y of 5 connected → (3,5)
#  x of 4 of y of 7 connected → (4,7)
#  x of 5 of y of 11 connected → (5,11) 
plt.scatter(x, y)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Scatter Plot')
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np  

sales = np.array([100000, 200000, 300000, 400000, 500000, 600000, 700000])
days = np.array(['Friday', 'Saturday', 'Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday'])

# Line plot
plt.plot(days, sales, marker='o', linestyle='-', color='b')  # marker point line style and color

plt.xlabel('Days')
plt.ylabel('Sales')
plt.title('Sales per Day')
plt.grid(True)
plt.show()
```
