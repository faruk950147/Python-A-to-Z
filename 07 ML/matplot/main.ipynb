matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 3, 5, 7, 9])
plt.figure(figsize=(6.5, 4))
plt.plot(x, y, color='red')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Line Plot Figure')
plt.grid()
plt.legend(['Line Plot'])
plt.show()  

cls = np.array([1, 2, 3, 4, 5])
students = np.array([2, 3, 5, 7, 9])
# plt.plot(cls, students) thats means cls of 1 will be connected with students of 2 → (1,2) 
# cls of 1 of students of 2 connected → (1,2)
# cls of 2 of students of 3 connected → (2,3)
# cls of 3 of students of 5 connected → (3,5)
# cls of 4 of students of 7 connected → (4,7)
# cls of 5 of students of 9 connected → (5,9)
plt.plot(cls, students, color='blue', marker='o', linestyle='--', linewidth=2, markersize=8)
plt.xlabel('Classes')
plt.ylabel('Students')
plt.title('Line Plot')
plt.grid()
plt.legend(['Students'])
plt.show()
```

```python
cls = np.array([1, 2, 3, 4, 5])
students = np.array([2, 3, 5, 7, 9])
# plt.scatter(cls, students) thats means cls of 1 will be connected with students of 2 → (1,2) 
# cls of 1 of students of 2 connected → (1,2)
# cls of 2 of students of 3 connected → (2,3)
# cls of 3 of students of 5 connected → (3,5)
# cls of 4 of students of 7 connected → (4,7)
# cls of 5 of students of 9 connected → (5,9)
plt.scatter(cls, students, color='red', s=100, alpha=0.5)
plt.xlabel('Classes')
plt.ylabel('Students')
plt.title('Scatter Plot')
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np  

sales = np.array([100000, 200000, 300000, 400000, 500000, 600000, 700000])
days = np.array(['Friday', 'Saturday', 'Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday'])

# Line plot
plt.plot(days, sales, marker='o', linestyle='-', color='b')  # marker point line style and color

plt.xlabel('Days')
plt.ylabel('Sales')
plt.title('Sales per Day')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np  

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 3, 5, 7, 10])

plt.plot(x, y, marker='o', linestyle='-', color='blue', linewidth=2, markersize=8)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('Line Plot')
plt.grid()

# x-axis and y-axis just integer 
plt.xticks(x)  # x-axis ticks to integer
plt.yticks(y)  # y-axis ticks to integer

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# data
x = np.arange(1, 11)
y = x**2

# main figure
fig = plt.figure()

# large plot
ax1 = fig.add_axes([0.1, 0.1, 0.8, 0.8])
ax1.plot(x, y, 'r')
ax1.set_title("Main Plot")

# small plot
ax2 = fig.add_axes([0.3, 0.3, 0.4, 0.4])
ax2.plot(x, np.sqrt(x), 'b')
ax2.set_title("smaller plot")

plt.show()
```
