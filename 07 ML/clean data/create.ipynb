```python
import pandas as pd
import random

# Sample messy values
names = [" Rahim ", "karim", "SALMA", "Jamal ", " nabila", "RAFI", None]
emails = ["rahim@gmail.com", "karim@gmail.com", "", "salma@gmail.com", None]
amounts = [500, 700, "800", "One Thousand", None, 1200, 450]
dates = ["01-03-2025", "2025/03/01", "March 1, 2025", None]

data = []

for i in range(1000):
    row = {
        "Customer": random.choice(names),
        "Email": random.choice(emails),
        "Amount": random.choice(amounts),
        "Date": random.choice(dates)
    }
    data.append(row)

df = pd.DataFrame(data)

# Save CSV file
df.to_csv("1000_messy_customer_sales_data.csv", index=False)

print("CSV file created successfully!")
```
