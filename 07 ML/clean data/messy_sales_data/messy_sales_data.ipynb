```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('messy_sales_data.csv')
# Convert Quantity to numeric
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
plt.plot(df['Quantity'])
plt.xlabel('Index')
plt.xscale('log')
plt.ylabel('Quantity')
plt.title('Quantity over Rows')
plt.show()
```
