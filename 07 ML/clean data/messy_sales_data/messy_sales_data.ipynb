```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('messy_sales_data_cleaned.csv')

cols = ['Quantity', 'Unit Price', 'Total Amount']
means = [df[col].mean() for col in cols]

plt.figure(figsize=(6,4))
plt.bar(cols, means, color='skyblue')
# <-- Log scale makes small bars visible
plt.yscale('log')  
plt.title('Mean Value per Column (Log Scale)')
plt.ylabel('Mean Value (log scale)')
plt.show()
```
