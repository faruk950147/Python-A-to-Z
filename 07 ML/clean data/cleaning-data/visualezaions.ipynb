```python
import pandas as pd
import matplotlib.pyplot as plt

# Cleaned CSV Loading
df = pd.read_csv("titanic_cleaned.csv")

# Survival distribution
plt.figure(figsize=(6,4))
df['Survived'].value_counts().plot(kind='bar', color=['lightcoral', 'lightgreen'])
plt.xticks([0,1], ['Died', 'Survived'], rotation=0)
plt.title("Survival Count")
plt.ylabel("Count")
plt.show()

# Survival by Sex
plt.figure(figsize=(6,4))
df[df['Sex']==0]['Survived'].value_counts().sort_index().plot(kind='bar', color='lightblue', position=0, width=0.4, label='Male')
df[df['Sex']==1]['Survived'].value_counts().sort_index().plot(kind='bar', color='pink', position=1, width=0.4, label='Female')
plt.xticks([0,1], ['Died','Survived'], rotation=0)
plt.title("Survival by Sex")
plt.legend()
plt.show()

# Survival by Pclass
plt.figure(figsize=(6,4))
for pclass in sorted(df['Pclass'].unique()):
    df[df['Pclass']==pclass]['Survived'].value_counts().sort_index().plot(kind='bar', alpha=0.6, label=f'Class {pclass}')
plt.xticks([0,1], ['Died','Survived'], rotation=0)
plt.title("Survival by Passenger Class")
plt.legend()
plt.show()

# Age distribution
plt.figure(figsize=(8,5))
plt.hist(df['Age'], bins=30, color='skyblue', edgecolor='black')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()

# Age vs Survival (stacked histogram)
plt.figure(figsize=(8,5))
plt.hist([df[df['Survived']==0]['Age'], df[df['Survived']==1]['Age']],
         bins=30, stacked=True, color=['lightcoral','lightgreen'], label=['Died','Survived'])
plt.title("Age Distribution by Survival")
plt.xlabel("Age")
plt.ylabel("Count")
plt.legend()
plt.show()

# Fare distribution
plt.figure(figsize=(8,5))
plt.hist(df['Fare'], bins=30, color='orange', edgecolor='black')
plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Count")
plt.show()

# Fare vs Survival (boxplot)
plt.figure(figsize=(6,4))
plt.boxplot([df[df['Survived']==0]['Fare'], df[df['Survived']==1]['Fare']],
            labels=['Died','Survived'])
plt.title("Fare by Survival")
plt.ylabel("Fare")
plt.show()

# Survival by Title
plt.figure(figsize=(6,4))
titles = df['Title'].unique()
counts = [df[(df['Title']==t) & (df['Survived']==1)].shape[0] for t in titles]
plt.bar([str(t) for t in titles], counts, color='lightgreen')
plt.title("Survival by Title")
plt.xlabel("Title")
plt.ylabel("Survived Count")
plt.show()

# Survival by Embarked_1 (C)
plt.figure(figsize=(6,4))
counts = [df[(df['Embarked_1']==False) & (df['Survived']==1)].shape[0],
          df[(df['Embarked_1']==True) & (df['Survived']==1)].shape[0]]
plt.bar(['Not C','C'], counts, color='lightblue')
plt.title("Survival from Embarked C")
plt.ylabel("Survived Count")
plt.show()

# Survival by Embarked_2 (Q)
plt.figure(figsize=(6,4))
counts = [df[(df['Embarked_2']==False) & (df['Survived']==1)].shape[0],
          df[(df['Embarked_2']==True) & (df['Survived']==1)].shape[0]]
plt.bar(['Not Q','Q'], counts, color='orange')
plt.title("Survival from Embarked Q")
plt.ylabel("Survived Count")
plt.show()

# Age vs Fare scatter colored by Survival
plt.figure(figsize=(8,5))
plt.scatter(df[df['Survived']==0]['Age'], df[df['Survived']==0]['Fare'], color='red', alpha=0.5, label='Died')
plt.scatter(df[df['Survived']==1]['Age'], df[df['Survived']==1]['Fare'], color='green', alpha=0.5, label='Survived')
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Fare vs Age by Survival")
plt.legend()
plt.show()
```
