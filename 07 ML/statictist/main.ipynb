```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("titanic.csv")
# Numeric column → mean
df['Age'].fillna(df['Age'].mean(), inplace=True)

# Categorical column → mode
df['Cabin'].fillna(df['Cabin'].mode()[0], inplace=True)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
```

```python
print(df['Age'].mean())
```

```python
print(np.mean(df['Age']))
```

```python
sns.histplot(x=df['Age'], kde=True, bins= [i for i in range(0, 81, 10)])
plt.show()
```

```python
mean_value = df['Age'].mean()

# Histogram
sns.histplot(df['Age'], kde=True, bins=[i for i in range(0, 81, 10)])

# Mean line
plt.axvline(x=mean_value, color='red',linestyle='--', label=f'Mean = {mean_value}')

plt.legend()
plt.show()
```

```python
print(df['Age'].median())

print(np.median(df['Age']))
```

```python
sns.histplot(x=df['Age'], kde=True, bins= [i for i in range(0, 81, 10)])
plt.show()
```

```python
median_value = df['Age'].median()

# Histogram
sns.histplot(df['Age'], kde=True, bins=[i for i in range(0, 81, 10)])

# Median line
plt.axvline(x=median_value, color='red',linestyle='--', label=f'Median = {median_value}')

plt.legend()
plt.show()
```
