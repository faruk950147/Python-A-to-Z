```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("titanic.csv")
```

```python
print(df['Age'].mean())
```

```python
print(np.mean(df['Age']))
```

```python
mn = np.mean(df['Age'])
sns.histplot(x=df['Age'], kde=True, bins= [i for i in range(0, 81, 10)])
plt.plot([mn for i in range(300)], [i for i in range(300)], color='red')
plt.show()
```
