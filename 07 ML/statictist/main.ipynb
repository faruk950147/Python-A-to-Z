```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("titanic.csv")
df['Age'].fillna(df['Age'].mean(), inplace=True)
```

```python
print(df['Age'].mean())
```

```python
print(np.mean(df['Age']))
```

```python
sns.histplot(x=df['Age'], kde=True, bins= [i for i in range(0, 81, 10)])
plt.show()
```

```python
mean = np.mean(df['Age'])
sns.histplot(df['Age'], kde=True, bins=[i for i in range(0, 81, 10)])
plt.plot([mean for i in range(len(df['Age']))], [i for i in range(len(df['Age']))], color='red')
plt.show()
```

```python
print(df['Age'].median())

print(np.median(df['Age']))
```

```python
sns.histplot(x=df['Age'], kde=True, bins= [i for i in range(0, 81, 10)])
plt.show()
```

```python
median = np.median(df['Age'])
sns.histplot(df['Age'], kde=True, bins=[i for i in range(0, 81, 10)])
plt.plot([median for i in range(len(df['Age']))], [i for i in range(len(df['Age']))], color='red')
plt.show()
```
