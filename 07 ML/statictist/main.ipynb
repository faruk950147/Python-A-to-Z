```python
import numpy as np
from scipy import stats

# 1 case
case_1 = [11, 9, 10, 14, 11, 8, 9, 12, 7, 8, 8, 9, 11, 14, 10, 9]

# 2 case
case_2 = [15, 13, 12, 13, 14, 13, 16, 19, 13, 14, 10, 16, 14, 15, 13, 14]

# 3 case
case_3 = [10, 10, 12, 11, 9, 8, 10, 9, 9, 11, 9, 11, 10, 10, 11, 10]

def calculate_statistics(data, case_name):
    print(f"--- {case_name} ---")
    
    # Mean
    mean = np.mean(data)
    
    # Median
    median = np.median(data)
    
    # Mode
    mode_res = stats.mode(data, keepdims=True)
    mode = mode_res.mode[0]
    
    # Standard Deviation (Population)
    std_dev = np.std(data, ddof=0)
    
    # Range
    data_range = np.max(data) - np.min(data)
    
    # result
    print(f"Mean                      : {mean:.2f}")
    print(f"Median               : {median:.1f}")
    print(f"Mode                 : {mode}")
    print(f"Standard Deviation   : {std_dev:.2f}")
    print(f"Range                   : {data_range}")
    print("\n")

# all calculations
calculate_statistics(case_1, "Case 1")
calculate_statistics(case_2, "Case 2")
calculate_statistics(case_3, "Case 3")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import numpy as np

data = np.array([2, 4, 6])
''' 
mean = Σx/n
mean = (2+4+6)/3 = 4
'''

print("Mean:", np.mean(data))

# MAD
print("MAD:", np.mean(np.abs(data - np.mean(data))))
'''
mad = Σ|x - mean|/n
mad = (2+0+2)/3 = 1.33

[2, 4, 6] - 4 = [-2, 0, 2]
abs([-2, 0, 2]) = [2, 0, 2]
(2+0+2)/3 = 1.333
'''

# Sample STD
print("STD:", np.std(data, ddof=1))
'''
std = √(Σ(x - mean)²/n-1)
std = √((4+0+4)/2) = √4 = 2

[2, 4, 6] - 4 = [-2, 0, 2]
(-2)² + (0)² + (2)² = 4 + 0 + 4 = 8
8/2 = 4
√4 = 2

'''

# Sample Variance
print("Variance:", np.var(data, ddof=1))
'''
variance = Σ(x - mean)²/n-1
variance = (4+0+4)/2 = 4

[2, 4, 6] - 4 = [-2, 0, 2]
(-2)² + (0)² + (2)² = 4 + 0 + 4 = 8
8/2 = 4
'''
```

```python
'''
========================= Percentiles, Quartiles, Quintiles and IQR =========================

1.  Percentiles

A percentile divides a dataset into 100 equal parts.

Examples: P25 → 25% of data values lie below it P50 → Median P90 → 90%
of data values lie below it

Ungrouped Data Formula: Position = k(n+1)/100

Where: Pk = k-th percentile k = percentile number n = total observations

Example: Data: 10, 20, 30, 40, 50

P50 = 50(5+1)/100 = 3 P50 = 30

Grouped Data Formula:

Pk = L + (((kN/100) - CF) / f) * h

Where: L = lower boundary N = total frequency CF = cumulative frequency
before percentile class f = frequency of percentile class h = class
width

Example (Grouped Data Percentile)

Class Interval | Frequency | Cumulative Frequency 0-10 | 5 | 5 10-20 | 8
| 13 20-30 | 12 | 25 30-40 | 10 | 35 40-50 | 5 | 40

N = 40

Find P40

Step 1: kN/100 = 40*40/100 = 16

Step 2: Percentile class = 20-30

L = 20 CF = 13 f = 12 h = 10

Step 3: P40 = 20 + ((16-13)/12) * 10 P40 = 22.5

Answer: P40 = 22.5

2.  Quartiles

Qk = k(n+1)/4

Q1 → 25% Q2 → 50% (Median) Q3 → 75%

Grouped Formula: Qk = L + (((kN/4) - CF) / f) * h

3.  Quintiles

Qk = k(n+1)/5

Q1 → 20% Q2 → 40% Q3 → 60% Q4 → 80%

4.  IQR (Interquartile Range)

IQR = Q3 - Q1

Example: Data: 10, 20, 30, 40, 50, 60, 70

Q1 = 20 Q3 = 60

IQR = 40

Outlier Rule:

Lower Limit = Q1 - 1.5(IQR) Upper Limit = Q3 + 1.5(IQR)

'''

```
