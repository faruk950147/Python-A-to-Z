====================== Measures of Frequency Distribution ======================
Measures of Frequency Distribution

Measures of Frequency Distribution are statistical methods used to organize, summarize, and display how often each value or class occurs in a dataset. They help us understand the pattern and distribution of data.

The main measures of frequency distribution are:

1. Frequency (f)

Frequency is the number of times a particular value occurs in a dataset.

Example:

Data: 2, 3, 3, 4, 5, 3, 6

Frequency of 3 = 3

2. Relative Frequency

Relative Frequency is the proportion of the total observations represented by a particular frequency.

Formula:

Relative Frequency = Frequency / Total Number of Observations

Example:

Data: 2, 3, 3, 4, 5, 3, 6

Frequency of 3 = 3

Total observations = 7

Relative Frequency = 3 / 7 = 0.429 (42.9%)

3. Percentage Frequency

Percentage Frequency expresses the relative frequency as a percentage.

Formula:

Percentage Frequency = (Frequency / Total Observations) × 100

Example:

Percentage Frequency = (3 / 7) × 100 = 42.9%

4. Cumulative Frequency

Cumulative Frequency is the running total of frequencies up to a particular value or class.

Example:

Value	Frequency	Cumulative Frequency
2	1	1
3	3	4
4	1	5
5	1	6
6	1	7
Comparison
Measure	Definition	Formula	Best Used When
Frequency	Number of occurrences	Count	Counting observations
Relative Frequency	Proportion of observations	f / N	Comparing frequencies
Percentage Frequency	Frequency as a percentage	(f / N) × 100	Presenting data in percentages
Cumulative Frequency	Running total of frequencies	Sum of frequencies	Finding cumulative totals
Advantages
Frequency
Easy to understand.
Shows how often values occur.
Relative Frequency
Allows comparison between datasets.
Expresses frequency as a proportion.
Percentage Frequency
Easy to interpret.
Commonly used in reports and charts.
Cumulative Frequency
Helps determine medians, quartiles, and percentiles.
Useful for cumulative analysis.
Disadvantages
Frequency
Does not show proportions.
Less useful for comparing datasets of different sizes.
Relative Frequency
Requires calculation.
May involve decimals.
Percentage Frequency
Can involve rounding errors.
Requires conversion to percentages.
Cumulative Frequency
Does not show individual frequencies clearly.
Less suitable for identifying exact frequencies.
Example

Data: 2, 3, 3, 4, 5, 3, 6

Frequency of 3 = 3
Relative Frequency = 3/7 = 0.429
Percentage Frequency = 42.9%
Cumulative Frequency = 1, 4, 5, 6, 7
When to Use
Frequency: To count how many times each value appears.
Relative Frequency: To compare frequencies across datasets.
Percentage Frequency: To express frequencies as percentages.
Cumulative Frequency: To calculate cumulative totals and locate median or quartiles.
Summary
Frequency = Number of Occurrences
Relative Frequency = Frequency / Total Observations
Percentage Frequency = (Frequency / Total Observations) × 100
Cumulative Frequency = Running Total of Frequencies
