```python
'''
=========================== Measures of Central Tendency ============================
Measures of Central Tendency are statistical measures used to determine the central or representative value of a dataset.

The three main measures of central tendency are:

1. Arithmetic Mean

The Arithmetic Mean is the average of all observations. It is calculated by dividing the sum of all values by the total number of values.

Formula:

Mean = ΣX / N

Where:

ΣX = Sum of all observations
N = Total number of observations

Example:

Data: 10, 20, 30, 40, 50

Mean = (10 + 20 + 30 + 40 + 50) / 5 = 150 / 5 = 30

Mean = 30

2. Median

Definition:
The median is the middle value of a dataset when the data is arranged in ascending or descending order.

Step 1:
Arrange the data in ascending (or descending) order.

--------------------------------------------------

Case 1: If the number of observations (N) is odd

Formula:
Median = Value at [(N + 1) / 2]th position

Example:
Data: 2, 4, 6, 8, 10

N = 5

Median Position = (5 + 1) / 2 = 3rd position

Median = 6

--------------------------------------------------

Case 2: If the number of observations (N) is even

Formula:
Median = (Value at (N / 2)th position + Value at ((N / 2) + 1)th position) / 2

Example:
Data: 2, 4, 6, 8

N = 4

Median = (Value at 2nd position + Value at 3rd position) / 2

       = (4 + 6) / 2
       = 5

--------------------------------------------------

3. Mode

The Mode is the value that occurs most frequently in a dataset.
Formula:

Mode = L + ((f1 - f0) / (2f1 - f0 - f2)) * h

Where:
L  = Lower limit of the modal class
f1 = Frequency of the modal class
f0 = Frequency of the class before the modal class
f2 = Frequency of the class after the modal class
h  = Class width

Example:
data = [1, 2, 2, 3, 4, 4, 4, 5, 6]

Frequency Table:
Value  Frequency
1      1
2      2
3      1
4      3
5      1
6      1

Ungrouped Data:
Mode = 4

Grouped Data Example:

Class   Frequency
1-2     3
3-4     4   (Modal Class)
5-6     2

L  = 3
f0 = 3
f1 = 4
f2 = 2
h  = 2

Mode = 3 + ((4 - 3) / (2*4 - 3 - 2)) * 2
     = 3 + (1 / 3) * 2
     = 3.67 (approx.)


Comparison

Mean - Average value - Formula: ΣX / N - Used when all data values are important.

Median - Middle value - Used when the data contains outliers.

Mode - Most frequent value - Used to find the most common value.

Advantages

Mean:

Easy to calculate.
Uses all observations.
Suitable for mathematical analysis.

Median:

Not affected much by outliers.
Best for skewed data.

'''
```

```python
# ==================== Measures of Central Tendency ====================
import numpy as np
from scipy import stats
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

marks = np.array([85, 92, 78, 90, 88, 95, 82, 89, 91, 91])

# Mean = ΣX / N
mean = sum(marks) / len(marks)
print(f"Manual Mean: {mean}")
print(f"NumPy Mean: {np.mean(marks)}")

# Median 
# if N is odd (n+1/2) 
# if N is even (n/2 + n/2+1)/2
sorted_marks = sorted(marks)
n = len(sorted_marks)

median = (sorted_marks[n//2 - 1] + sorted_marks[n//2]) / 2

print(f"Manual Median: {median}")
print(f"NumPy Median: {np.median(marks)}")

# Mode
print(f"SciPy Mode: {stats.mode(marks)}")
print(f"Pandas Mode: {pd.Series(marks).mode()}")
```
