============================ Population (A to Z) ============================

1. What is Population?

A Population is the complete set of all individuals, objects, events, or observations that a researcher wants to study.

In statistics, the population includes every member of the group of interest.

------------------------------------------------------------------

2. Definition

A Population is the complete collection of all individuals, objects, or observations of interest in a statistical study.

------------------------------------------------------------------

3. Symbols

Population Size = N

Population Mean = μ (Mu)

Population Variance = σ² (Sigma Square)

Population Standard Deviation = σ (Sigma)

------------------------------------------------------------------

4. Characteristics of Population

• Includes all members of the group.
• Can be finite or infinite.
• Used as the basis for statistical analysis.
• Described by parameters.
• May consist of people, animals, objects, or events.

------------------------------------------------------------------

5. Types of Population

A. Finite Population
A population with a limited number of members.

Example:
All 500 students in a college.

--------------------------------------------------

B. Infinite Population
A population with an unlimited number of members.

Example:
All possible coin tosses.
Continuous rainfall measurements.

--------------------------------------------------

C. Existing Population
A population that currently exists.

Example:
All employees in a company.

--------------------------------------------------

D. Hypothetical Population
A population that is imagined or theoretical.

Example:
All possible outcomes of rolling a die forever.

--------------------------------------------------

E. Target Population
The entire group the researcher wants to study.

Example:
All university students in Bangladesh.

--------------------------------------------------

F. Accessible Population
The part of the target population that the researcher can actually reach.

Example:
Students from one university selected for the study.

------------------------------------------------------------------

6. Population Size

Population Size is the total number of members in the population.

Symbol:
N

Example:
If a school has 2,000 students,

Population Size = N = 2000

------------------------------------------------------------------

7. Population Mean

The average of all values in the population.

Symbol:
μ

Formula:

μ = Σx / N

Where:

μ = Population Mean

Σx = Sum of all values

N = Population Size

Example:

Marks = 60, 70, 80, 90, 100

Σx = 60 + 70 + 80 + 90 + 100 = 400

N = 5

μ = 400 / 5 = 80

Population Mean = 80

------------------------------------------------------------------

8. Population Variance

Population Variance measures how far the population values are spread from the mean.

Symbol:
σ²

Formula:

σ² = Σ(x − μ)² / N

Example:

Marks = 60, 70, 80, 90, 100

μ = 80

Σ(x − μ)² = (60 - 80)² + (70 - 80)² + (80 - 80)² + (90 - 80)² + (100 - 80)² = 1000

N = 5

σ² = 1000 / 5 = 200

Population Variance = 200

------------------------------------------------------------------

9. Population Standard Deviation

Population Standard Deviation measures the average distance of values from the population mean.

Symbol:
σ

Formula:

σ = √σ²

Example:

σ = √200 = 14.14

Population Standard Deviation = 14.14

------------------------------------------------------------------

10. Population Parameters

Population characteristics are called Parameters.

Examples:

• Population Mean (μ)
• Population Variance (σ²)
• Population Standard Deviation (σ)
• Population Proportion (P)

------------------------------------------------------------------

11. Population vs Sample

Population
• Entire group
• Symbol = N
• Mean = μ
• More accurate
• Expensive
• Time-consuming

Sample
• Part of the population
• Symbol = n
• Mean = x̄
• Less accurate
• Less expensive
• Faster

13. Advantages of Population

• Provides complete information.

------------------------------------------------------------------

14. Disadvantages of Population

------------------------------------------------------------------

15. Key Points

• Population means the entire group being studied.
• Every member is included in the population.
• Population Size is represented by N.
• Population Mean is represented by μ.
• Population Variance is represented by σ².
• Population Standard Deviation is represented by σ.
• Population characteristics are called Parameters.

------------------------------------------------------------------

16. Frequently Asked Questions (FAQ)

Q1. What is Population?
A Population is the complete set of all individuals, objects, or observations of interest in a study.

Q2. What is the symbol for Population Size?
N

Q3. What is the symbol for Population Mean?
μ (Mu)

Q4. What is the formula for Population Mean?
μ = Σx / N

Q5. What is the difference between Population and Sample?
A Population includes every member of the group, while a Sample is only a subset of the population.

Q6. What are the main types of Population?
Finite, Infinite, Existing, Hypothetical, Target, and Accessible Population.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ============================ Population Statistics ============================

# Dataset
marks = np.array([60, 70, 80, 90, 100])

# ------------------------------------------------------------------------------
# Population Mean (μ)
# Formula: μ = Σx / N
# ------------------------------------------------------------------------------
population_mean = np.sum(marks) / len(marks)

print(f"Population Mean (μ): {population_mean:.2f}")
print(f"NumPy Mean: {np.mean(marks):.2f}")

# ------------------------------------------------------------------------------
# Population Variance (σ²)
# Formula: σ² = Σ(x − μ)² / N
# ------------------------------------------------------------------------------
population_variance = np.sum((marks - population_mean) ** 2) / len(marks)

print(f"Population Variance (σ²): {population_variance:.2f}")
print(f"NumPy Variance: {np.var(marks):.2f}")

# ------------------------------------------------------------------------------
# Population Standard Deviation (σ)
# Formula: σ = √σ²
# ------------------------------------------------------------------------------
population_std = np.sqrt(population_variance)

print(f"Population Standard Deviation (σ): {population_std:.2f}")
print(f"NumPy Standard Deviation: {np.std(marks):.2f}")
```

============================= Sample Statistics ============================

1. What is Sample?

A Sample is a subset or smaller portion of a population selected for statistical analysis.

Instead of studying the entire population, researchers study a sample to make conclusions about the population.

------------------------------------------------------------------

2. Definition

A Sample is a part of the population chosen to represent the entire population.

------------------------------------------------------------------

3. Symbols

Sample Size = n

Sample Mean = x̄ (x-bar)

Sample Variance = s²

Sample Standard Deviation = s

------------------------------------------------------------------

4. Characteristics of Sample

• Represents the population.
• Smaller than the population.
• Saves time and cost.
• Easier to collect data.
• Used to estimate population parameters.

------------------------------------------------------------------

5. Types of Sample

A. Random Sample
Every member of the population has an equal chance of being selected.

Example:
Randomly selecting 100 students from a university.

--------------------------------------------------

B. Simple Random Sample
Each possible sample has an equal probability of being selected.

--------------------------------------------------

C. Systematic Sample
Every kth member is selected after a random starting point.

Example:
Selecting every 10th customer entering a store.

--------------------------------------------------

D. Stratified Sample
The population is divided into groups (strata), and samples are selected from each group.

Example:
Selecting students from each academic department.

--------------------------------------------------

E. Cluster Sample
The population is divided into clusters, and entire clusters are selected.

Example:
Selecting several schools and surveying all students in those schools.

--------------------------------------------------

F. Convenience Sample
Samples are selected based on ease of access.

Example:
Interviewing people at a nearby shopping mall.

------------------------------------------------------------------

6. Sample Size

Sample Size is the total number of observations in the sample.

Symbol:
n

Example:
If 50 students are selected from a college,

Sample Size = n = 50

------------------------------------------------------------------

7. Sample Mean

The average of all values in the sample.

Symbol:
x̄ (x-bar)

Formula:

x̄ = Σx / n

Where:

x̄ = Sample Mean

Σx = Sum of all sample values

n = Sample Size

Example:

Sample values = 65, 70, 75, 80, 85

Σx = 375

n = 5

x̄ = 375 / 5 = 75

Sample Mean = 75

------------------------------------------------------------------

8. Sample Variance

Sample Variance measures how much the sample values vary from the sample mean.

Symbol:
s²

Formula:

s² = Σ(x − x̄)² / (n − 1)

------------------------------------------------------------------

9. Sample Standard Deviation

Sample Standard Deviation measures the average spread of sample values around the sample mean.

Symbol:
s

Formula:

s = √s²

------------------------------------------------------------------

10. Sample Statistics

Sample characteristics are called Statistics.

Examples:

• Sample Mean (x̄)
• Sample Variance (s²)
• Sample Standard Deviation (s)
• Sample Proportion (p̂)

------------------------------------------------------------------

11. Population vs Sample

Population
• Entire group
• Symbol = N
• Mean = μ
• Variance = σ²
• Standard Deviation = σ
• Parameters
• More expensive
• More time-consuming

Sample
• Part of the population
• Symbol = n
• Mean = x̄
• Variance = s²
• Standard Deviation = s
• Statistics
• Less expensive
• Faster

------------------------------------------------------------------

15. Key Points

16. Frequently Asked Questions (FAQ)

Q1. What is a Sample?
A Sample is a subset of the population selected for analysis.

Q2. What is the symbol for Sample Size?
n

Q3. What is the symbol for Sample Mean?
x̄ (x-bar)

Q4. What is the formula for Sample Mean?
x̄ = Σx / n

Q5. What is the formula for Sample Variance?
s² = Σ(x − x̄)² / (n − 1)

Q6. What is the difference between Population and Sample?
A Population includes every member of the group, while a Sample includes only a subset selected from that group.

Q7. Why is a Sample used?
A Sample is used to save time, reduce cost, and make statistical inferences about the population.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ============================ Sample Statistics ============================

# Dataset
marks = np.array([60, 70, 80, 90, 100])

# ------------------------------------------------------------------------------
# Sample Mean (x̄)
# Formula: x̄ = Σx / n
# ------------------------------------------------------------------------------
sample_mean = np.sum(marks) / len(marks)

print(f"Sample Mean (x̄): {sample_mean:.2f}")
print(f"NumPy Mean: {np.mean(marks):.2f}")

# ------------------------------------------------------------------------------
# Sample Variance (s²)
# Formula: s² = Σ(x − x̄)² / (n − 1)
# ------------------------------------------------------------------------------
sample_variance = np.sum((marks - sample_mean) ** 2) / (len(marks) - 1)

print(f"Sample Variance (s²): {sample_variance:.2f}")
print(f"NumPy Variance: {np.var(marks, ddof=1):.2f}")

# ------------------------------------------------------------------------------
# Sample Standard Deviation (s)
# Formula: s = √s²
# ------------------------------------------------------------------------------
sample_std = np.sqrt(sample_variance)

print(f"Sample Standard Deviation (s): {sample_std:.2f}")
print(f"NumPy Standard Deviation: {np.std(marks, ddof=1):.2f}")
```
