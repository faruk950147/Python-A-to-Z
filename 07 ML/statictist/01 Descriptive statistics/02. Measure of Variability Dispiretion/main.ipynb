=========================== Measures of Variability or Dispiretion ============================

Measures of Variability are statistical measures that describe how spread out or dispersed the data values are around the central value. They help us understand the consistency and variation within a dataset.

The main measures of variability are:

1. Range

The Range is the difference between the highest and the lowest values in a dataset.

Formula:

Range = Maximum Value − Minimum Value
Mid-Range = (Maximum Value + Minimum Value) / 2

Example:

Data: 5, 8, 12, 15, 20

Range = 20 − 5 = 15

2. Variance

Variance measures the average squared deviation of each observation from the mean.

Formula (Population):

Variance (σ²) = Σ(X − μ)² / N

Where:

X = Individual observation
μ = Population mean
N = Number of observations

A larger variance indicates greater variability in the data.

3. Standard Deviation

Standard Deviation is the square root of the variance. It measures the average distance of data values from the mean.

Formula (Population):

Standard Deviation (σ) = √Variance

or

σ = √[Σ(X − μ)² / N]

A smaller standard deviation indicates that the data points are closer to the mean.

4. Interquartile Range (IQR)

The Interquartile Range measures the spread of the middle 50% of the data.

Formula:

IQR = Q3 − Q1
Example:
    Data:
    [50, 55, 60, 65, 68, 55, 68, 100]

    Step 1: Arrange in ascending order:
    [50, 55, 55, 60, 65, 68, 68, 100]

    Step 2: Find Median (Q2):
    Q2 = (60 + 65) / 2 = 62.5

    Step 3: Find Q1:
    Lower half = [50, 55, 55, 60]
    Q1 = (55 + 55) / 2 = 55

    Step 4: Find Q3:
    Upper half = [65, 68, 68, 100]
    Q3 = (68 + 68) / 2 = 68

    Step 5: Find IQR:
    IQR = Q3 − Q1
    IQR = 68 − 55 = 13

    Step 6: Find outliers:
    Lower bound = 55 - 1.5 * 13 = 35.5
    Upper bound = 68 + 1.5 * 13 = 87.5

    Answer:
    Q1 = 55
    Q2 = 62.5
    Q3 = 68
    IQR = 13
    Lower bound = 35.5
    Upper bound = 87.5
    35.5 or less is Lower Outlier.
    87.5 or more is Upper Outlier.
    35.5 to 87.5 is Normal data in this range.


Where:

Q1 = First Quartile (25th percentile)
Q3 = Third Quartile (75th percentile)

IQR is less affected by outliers than the range.

Comparison
Measure	Definition	Formula	Best Used When
Range	Difference between highest and lowest values	Max − Min	Quick measure of spread
Variance	Average squared deviation from the mean	Σ(X − μ)² / N	Statistical analysis
Standard Deviation	Square root of variance	√Variance	Measuring data dispersion
IQR	Spread of the middle 50% of data	Q3 − Q1	Data contains outliers
Advantages
Range
Easy to calculate.
Quick measure of dispersion.
Variance
Uses all observations.
Useful in advanced statistical analysis.
Standard Deviation
Easy to interpret.
Widely used in statistics and machine learning.
IQR
Resistant to outliers.
Suitable for skewed data.
Disadvantages
Range
Affected by extreme values.
Uses only two observations.
Variance
Measured in squared units.
Harder to interpret.
Standard Deviation
Affected by outliers.
Requires more calculations.
IQR
Ignores 50% of the data.
Less useful for mathematical analysis.
Example

Data: 2, 4, 6, 8, 10

Range = 10 − 2 = 8
Mean = 6
Variance = 8
Standard Deviation = √8 ≈ 2.83
Q1 = 4
Q3 = 8
IQR = 8 − 4 = 4
When to Use
Range: For a quick measure of spread.
Variance: For statistical and mathematical analysis.
Standard Deviation: To measure how far data values are from the mean.
IQR: When the dataset contains outliers.
Summary
Range = Maximum − Minimum
Variance = Average Squared Deviation
Standard Deviation = √Variance
IQR = Q3 − Q1
