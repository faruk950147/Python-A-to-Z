=========================== Measures of Variability or Dispiretion ============================

Measures of Variability are statistical measures that describe how spread out or dispersed the data values are around the central value. They help us understand the consistency and variation within a dataset.

The main measures of variability are:

1. Range

The Range is the difference between the highest and the lowest values in a dataset.

Formula:

Range = Maximum Value − Minimum Value
Mid-Range = (Maximum Value + Minimum Value) / 2

Example:

Data: 5, 8, 12, 15, 20

Range = 20 − 5 = 15

2. Variance

Variance measures the average squared deviation of each observation from the mean.

Formula (Population):

Variance (σ²) = Σ(X − μ)² / N

Where:

X = Individual observation
μ = Population mean
N = Number of observations

A larger variance indicates greater variability in the data.

3. Standard Deviation

Standard Deviation is the square root of the variance. It measures the average distance of data values from the mean.

Formula (Population):

Standard Deviation (σ) = √Variance

or

σ = √[Σ(X − μ)² / N]

A smaller standard deviation indicates that the data points are closer to the mean.

4. Interquartile Range (IQR)

The Interquartile Range measures the spread of the middle 50% of the data.

Formula:

IQR = Q3 − Q1
Example:
    Data:
    [50, 55, 60, 65, 68, 55, 68, 100]

    Step 1: Arrange in ascending order:
    [50, 55, 55, 60, 65, 68, 68, 100]

    Step 2: Find Median (Q2):
    Q2 = (60 + 65) / 2 = 62.5

    Step 3: Find Q1:
    Lower half = [50, 55, 55, 60]
    Q1 = (55 + 55) / 2 = 55

    Step 4: Find Q3:
    Upper half = [65, 68, 68, 100]
    Q3 = (68 + 68) / 2 = 68

    Step 5: Find IQR:
    IQR = Q3 − Q1
    IQR = 68 − 55 = 13

    Step 6: Find outliers:
    Lower bound = 55 - 1.5 * 13 = 35.5
    Upper bound = 68 + 1.5 * 13 = 87.5

    Step 7: Identify outliers:
    100 is greater than 87.5, so it is an upper outlier.

    Step 8:
    mean = 66.25
    variance = 426.5625
    standard_deviation = 20.6534

    Answer:
    Q1 = 55
    Q2 = 62.5
    Q3 = 68
    IQR = 13
    Lower bound = 35.5
    Upper bound = 87.5
    35.5 or less is Lower Outlier.
    87.5 or more is Upper Outlier.
    35.5 to 87.5 is Normal data in this range.


Where:

Q1 = First Quartile (25th percentile)
Q3 = Third Quartile (75th percentile)

IQR is less affected by outliers than the range.


```python
# =========================== Measures of Variability (Dispersion) ===========================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# =============================== Range =====================================
# Formula:
# Range = Maximum Value - Minimum Value
# Mid_Range = (Maximum Value + Minimum Value) / 2


marks = np.array([2, 4, 6, 8, 10])

print("Range =", np.max(marks) - np.min(marks))
print("Mid-Range =", (np.max(marks) + np.min(marks)) / 2)


# ============================= Population Statistics =============================
# ------------------------------------------------------------------------------
# Population Mean (μ)
# Formula: μ = Σx / N
# ------------------------------------------------------------------------------
population_mean = np.sum(marks) / len(marks)

print(f"Population Mean (μ): {population_mean:.2f}")
print(f"NumPy Mean: {np.mean(marks):.2f}")

# ------------------------------------------------------------------------------
# Population Variance (σ²)
# Formula: σ² = Σ(x − μ)² / N
# ------------------------------------------------------------------------------
population_variance = np.sum((marks - population_mean) ** 2) / len(marks)

print(f"Population Variance (σ²): {population_variance:.2f}")
print(f"NumPy Variance: {np.var(marks):.2f}")

# ------------------------------------------------------------------------------
# Population Standard Deviation (σ)
# Formula: σ = √σ²
# ------------------------------------------------------------------------------
population_std = np.sqrt(population_variance)

print(f"Population Standard Deviation (σ): {population_std:.2f}")
print(f"NumPy Standard Deviation: {np.std(marks):.2f}")


# ============================ Sample Statistics ============================
# ------------------------------------------------------------------------------
# Sample Mean (x̄)
# Formula: x̄ = Σx / n
# ------------------------------------------------------------------------------
sample_mean = np.sum(marks) / len(marks)

print(f"Sample Mean (x̄): {sample_mean:.2f}")
print(f"NumPy Mean: {np.mean(marks):.2f}")

# ------------------------------------------------------------------------------
# Sample Variance (s²)
# Formula: s² = Σ(x − x̄)² / (n − 1)
# ------------------------------------------------------------------------------
sample_variance = np.sum((marks - sample_mean) ** 2) / (len(marks) - 1)

print(f"Sample Variance (s²): {sample_variance:.2f}")
print(f"NumPy Variance: {np.var(marks, ddof=1):.2f}")

# ------------------------------------------------------------------------------
# Sample Standard Deviation (s)
# Formula: s = √s²
# ------------------------------------------------------------------------------
sample_std = np.sqrt(sample_variance)

print(f"Sample Standard Deviation (s): {sample_std:.2f}")
print(f"NumPy Standard Deviation: {np.std(marks, ddof=1):.2f}")

# 4. Coefficient of Variation (CV)
# Definition:
# The Coefficient of Variation (CV) is a standardized measure of dispersion that expresses the standard deviation as a percentage of the mean.
# It is useful for comparing the variability of datasets with different units or scales.

# Formula:
# CV = (σ / μ) × 100%
# data = np.array([2, 4, 6, 8, 10])

cv = (population_std / population_mean) * 100

print("Coefficient of Variation (CV) =", cv)


# 5. Interquartile Range (IQR)
# Definition:
# The Interquartile Range (IQR) measures the spread of the middle 50% of the data.

# Formula:
# IQR = Q3 - Q1
# data = np.array([2, 4, 6, 8, 10])

q1 = np.percentile(marks, 25)
q3 = np.percentile(marks, 75)

iqr = q3 - q1

print("Q1 =", q1)
print("Q3 =", q3)
print("IQR =", iqr)

```

```python
import numpy as np

# Data
data = np.array([50, 55, 60, 65, 68, 55, 68, 100])

# Step 1: Arrange data in ascending order
sorted_data = np.sort(data)

# Step 2: Find Quartiles
q1 = np.percentile(sorted_data, 25)
q2 = np.median(sorted_data)
q3 = np.percentile(sorted_data, 75)

# Step 3: Find IQR
iqr = q3 - q1

# Step 4: Find Outlier Boundaries
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

# Step 5: Identify Outliers
lower_outliers = sorted_data[sorted_data < lower_bound]
upper_outliers = sorted_data[sorted_data > upper_bound]
normal_data = sorted_data[
    (sorted_data >= lower_bound) & (sorted_data <= upper_bound)
]

# Step 6: Mean, Variance, Standard Deviation
mean = np.sum(sorted_data) / len(sorted_data)
variance = np.sum((sorted_data - mean) ** 2) / len(sorted_data)
std = np.sqrt(variance)

# Output
print("\nQ1 =", q1)
print("Q2 =", q2)
print("Q3 =", q3)
print("IQR =", iqr)

print("\nLower Bound =", lower_bound)
print("Upper Bound =", upper_bound)

print("\nLower Outliers:", lower_outliers)
print("Upper Outliers:", upper_outliers)
print("\nMean =", mean)
print("Variance =", variance)
print("Standard Deviation =", round(std, 4))
```
