============================ Inferential Statistics ============================

1. What is Inferential Statistics?

Inferential Statistics is the branch of statistics that uses sample data to make conclusions, predictions, or decisions about an entire population.

Instead of studying every member of the population, inferential statistics analyzes a sample and draws inferences about the population.

------------------------------------------------------------------

2. Definition

Inferential Statistics is the process of using sample data to estimate population parameters and test hypotheses.

------------------------------------------------------------------

3. Purpose of Inferential Statistics

• To make predictions about a population.
• To estimate unknown population parameters.
• To test research hypotheses.
• To support decision-making using sample data.
• To measure uncertainty and confidence.

------------------------------------------------------------------

4. Characteristics

• Based on sample data.
• Uses probability theory.
• Makes generalizations about a population.
• Involves estimation and hypothesis testing.
• Results are subject to sampling error.

------------------------------------------------------------------

5. Main Branches of Inferential Statistics

A. Estimation

Estimation is used to estimate unknown population parameters from sample statistics.

Types:
• Point Estimation
• Interval Estimation

Example:
Estimating the average income of all citizens using a sample.

--------------------------------------------------

B. Hypothesis Testing

Hypothesis testing is used to determine whether a claim about a population is true.

Types:
• Null Hypothesis (H₀)
• Alternative Hypothesis (H₁ or Ha)

Example:
Testing whether the average exam score is greater than 70.

------------------------------------------------------------------

6. Common Statistical Tests

• Z-Test
• t-Test
• Chi-Square Test
• F-Test
• ANOVA (Analysis of Variance)

------------------------------------------------------------------

7. Important Terms

Population = Entire group

Sample = Part of the population

Parameter = Population characteristic

Statistic = Sample characteristic

Confidence Level = Degree of certainty (e.g., 95%)

Margin of Error = Maximum expected difference between the sample result and the true population value

P-Value = Probability of obtaining the observed result if the null hypothesis is true

------------------------------------------------------------------

8. Steps in Inferential Statistics

Step 1:
Define the research problem.

Step 2:
Select a representative sample.

Step 3:
Collect sample data.

Step 4:
Analyze the data.

Step 5:
Estimate population parameters or test hypotheses.

Step 6:
Draw conclusions about the population.

------------------------------------------------------------------

9. Real-Life Examples

• Predicting election results using voter surveys.
• Estimating the average salary of employees.
• Testing whether a new medicine is effective.
• Estimating customer satisfaction from survey responses.
• Predicting product demand using market research.

------------------------------------------------------------------

10. Advantages

• Saves time and money.
• Suitable for large populations.
• Supports scientific decision-making.
• Allows prediction of future outcomes.
• Helps estimate unknown population values.

------------------------------------------------------------------

11. Disadvantages

• Results may contain sampling errors.
• Accuracy depends on sample quality.
• Incorrect sampling may produce biased results.
• Conclusions are probabilistic, not certain.

------------------------------------------------------------------

12. Descriptive Statistics vs Inferential Statistics

Descriptive Statistics
• Summarizes collected data.
• Does not make predictions.
• Describes only observed data.

Inferential Statistics
• Uses sample data.
• Makes predictions about the population.
• Tests hypotheses.
• Estimates population parameters.
