```python
# ============================== Probability ==============================

# What is Probability?
"""
PROBABILITY FULL NOTES

1. Definition

Probability measures how likely an event is to occur.

0 ≤ P(E) ≤ 1

0 = Impossible event

1 = Certain event

2. Basic Terms

Experiment

An action that produces outcomes.

Examples: Tossing a coin, rolling a die.

Outcome

A possible result of an experiment.

Example: Die → 1, 2, 3, 4, 5, 6

Sample Space (S)

Set of all possible outcomes.

Coin: S = {H, T}

Die: S = {1, 2, 3, 4, 5, 6}

Event (E)

A subset of the sample space.

Example: Even numbers = {2, 4, 6}

3. Probability Formula

P(E) = (Favorable Outcomes) / (Total Outcomes)

Example: Probability of getting 5 on a die = 1/6

4. Types of Events

Simple Event

Contains one outcome.

Example: Getting 4.

Compound Event

Contains more than one outcome.

Example: Getting an even number.

Impossible Event

Cannot occur.

Example: Getting 8 on a die.

P = 0

Certain Event

Must occur.

Example: Getting a number less than 7 on a die.

P = 1

5. Complementary Event

P(E') = 1 - P(E)

Example: P(Odd) = 3/6 = 1/2

P(Not Odd) = 1 - 1/2 = 1/2

6. Addition Rule

P(A ∪ B) = P(A) + P(B) - P(A ∩ B)

Example:

A = {2,4,6}

B = {2,3,5}

P(A)=3/6, P(B)=3/6, P(A∩B)=1/6

P(A ∪ B)=3/6 + 3/6 - 1/6 = 5/6

7. Conditional Probability

P(A|B) = P(A ∩ B) / P(B)

Probability of A occurring given B has occurred.

8. Independent Events

P(A ∩ B) = P(A) * P(B)

Example:

P(Head)=1/2

P(6)=1/6

P(Head and 6)=1/2 * 1/6 = 1/12

9. Dependent Events

One event affects the other.

Example: Drawing cards without replacement.

10. Bayes' Theorem

P(A|B) = [P(B|A) * P(A)] / P(B)

11. Counting Principle

If one task has m ways and another has n ways, total ways = m * n.

12. Permutation (Order Matters)

nPr = n! / (n-r)!

Example: 5P3 = 60

13. Combination (Order Does Not Matter)

nCr = n! / [r!(n-r)!]

Example: 5C3 = 10

14. Random Variable

A variable whose value depends on chance.

15. Probability Distribution

Example:

X : 0, 1, 2

P(X): 0.2, 0.5, 0.3

Sum of probabilities = 1

16. Expected Value

E(X) = Σ xP(x)

17. Properties of Probability

0 ≤ P(E) ≤ 1

P(S) = 1

P(∅) = 0

P(E') = 1 - P(E)

P(A ∪ B) = P(A) + P(B) - P(A ∩ B)

18. Important Formulas

| Concept | Formula |
| Probability | P(E)=n(E)/n(S) |

| Complement | P(E')=1-P(E) |

| Addition Rule | P(A∪B)=P(A)+P(B)-P(A∩B) |

| Conditional Probability | P(A|B)=P(A∩B)/P(B) |

| Independent Events | P(A∩B)=P(A)P(B) |

| Bayes' Theorem | P(A|B)=[P(B|A)P(A)]/P(B) |

| Permutation | nPr=n!/(n-r)! |

| Combination | nCr=n!/[r!(n-r)!] |

| Expected Value | E(X)=ΣxP(x) |

Quick Tricks

Permutation → Order matters (Arrange)

Combination → Order does not matter (Choose)

Independent events → Multiply

Mutually exclusive events → Add

Complement → 1 - Probability
"""
```
