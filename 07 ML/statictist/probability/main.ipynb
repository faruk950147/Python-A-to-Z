```python
# ================================== Probability ==================================

"""
What is Probability?
--------------------
Probability is a branch of mathematics that measures the likelihood (chance)
of an event occurring.

The value of probability always lies between 0 and 1.

- P(E) = 0  → Impossible Event
- P(E) = 1  → Certain Event
- 0 < P(E) < 1 → Possible Event

Formula
-------
P(E) = Number of Favorable Outcomes / Total Number of Possible Outcomes

or,

             Favorable Outcomes
P(E) = -------------------------------
        Total Possible Outcomes

Where:
- P(E) = Probability of event E

Example:
--------
If a die is rolled,

Total possible outcomes = 6
Favorable outcomes for getting 4 = 1

P(Getting 4) = 1/6


====================================
Types of Probability
====================================

1. Theoretical Probability
--------------------------
Theoretical Probability is calculated using mathematical reasoning,
without performing an experiment.

Formula:
P(E) = Favorable Outcomes / Total Possible Outcomes

Examples:
- Probability of getting Head when tossing a coin = 1/2
- Probability of getting 6 when rolling a die = 1/6


2. Experimental Probability
---------------------------
Experimental Probability is calculated using the actual results
obtained from performing an experiment.

Formula:
P(E) = Number of Times Event Occurred / Total Number of Trials

Example:
A coin is tossed 100 times.

Head appears = 56 times

P(Head) = 56/100 = 0.56


3. Axiomatic Probability
------------------------
Axiomatic Probability is based on probability rules (axioms)
introduced by Russian mathematician Andrey Kolmogorov.

The three axioms are:

1. Probability is never negative.
   P(E) ≥ 0

2. The probability of the sample space is 1.
   P(S) = 1

3. For mutually exclusive events:
   P(A ∪ B) = P(A) + P(B)

Example:
- Probability of getting a Head from a fair coin = 1/2
- Probability of getting a 6 from a fair die = 1/6


====================================
Important Terms
====================================

Sample Space (S):
The set of all possible outcomes.

Example:
Coin → {H, T}
Die → {1, 2, 3, 4, 5, 6}

Event (E):
A subset of the sample space.

Example:
Getting an even number from a die:
E = {2, 4, 6}

Favorable Outcome:
The outcomes that satisfy the event.

Example:
Event = Getting a number greater than 4
Favorable outcomes = {5, 6}


====================================
Probability Properties
====================================

1. 0 ≤ P(E) ≤ 1

2. P(Impossible Event) = 0

3. P(Certain Event) = 1

4. Complement Rule:
   P(E') = 1 − P(E)

5. Addition Rule:
   P(A ∪ B) = P(A) + P(B) − P(A ∩ B)

6. Multiplication Rule (Independent Events):
   P(A ∩ B) = P(A) × P(B)


====================================
Real-Life Applications
====================================

- Weather forecasting
- Machine Learning
- Artificial Intelligence
- Medical diagnosis
- Stock market analysis
- Risk assessment
- Insurance
- Data Science
- Games and gambling
- Quality control
"""
```
