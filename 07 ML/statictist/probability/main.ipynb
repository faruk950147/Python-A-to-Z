# ============================== Probability ==============================

# What is Probability?
"""
Probability is a measure of the likelihood of an event occurring.

It is a numerical value between 0 and 1:
- 0 means the event is impossible.
- 1 means the event is certain.

Formula:
Probability = (Number of Favorable Outcomes) / (Total Number of Possible Outcomes)

P(E) = Number of Favorable Outcomes / Total Number of Possible Outcomes
"""

# Example 1: Tossing a Coin
"""
Sample Space = {Head, Tail}

Probability of getting a Head:
P(Head) = 1 / 2 = 0.5

Probability of getting a Tail:
P(Tail) = 1 / 2 = 0.5
"""

# Example 2: Rolling a Die
"""
Sample Space = {1, 2, 3, 4, 5, 6}

Probability of getting a 4:
P(4) = 1 / 6

Probability of getting an even number:
Even Numbers = {2, 4, 6}

P(Even) = 3 / 6 = 1 / 2 = 0.5
"""

# Key Points
"""
- Probability ranges from 0 to 1.
- 0 → Impossible event
- 1 → Certain event
- 0.5 → Equally likely event
"""

# Real-Life Examples
"""
- Probability of the sun rising tomorrow = 1
- Probability of rolling a 7 on a standard die = 0
- Probability of drawing an Ace from a deck of 52 cards = 4/52 = 1/13
"""
