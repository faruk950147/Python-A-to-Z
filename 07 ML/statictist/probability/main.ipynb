```python
# ================================= Probability ==========================================
"""
    # 1. Introduction to Probability

        ## What is Probability?

        Probability is a branch of mathematics that measures the likelihood (chance) of an event occurring.

        The value of probability always lies between 0 and 1.

        ## Interpretation of Probability Values

        - P(E) = 0 → Impossible Event (The event can never occur.)
        - P(E) = 1 → Certain Event (The event will definitely occur.)
        - 0 < P(E) < 1 → Possible Event (The event may occur.)
        - P(E) = 0.5 → The event has a 50% chance of occurring.

        --------------------------------------------------

        ## Types of Probability

        ### 1. Theoretical Probability

        Theoretical Probability is calculated using mathematical reasoning without performing an experiment.

        **Formula:**

        P(E) = Number of Favorable Outcomes / Total Number of Possible Outcomes

        or

        P(E) = n(E) / n(S)

        Where:
        - P(E) = Probability of event E
        - n(E) = Number of favorable outcomes
        - n(S) = Total number of outcomes in the sample space

        **Examples:**
        - Probability of getting Head when tossing a coin = 1/2 = 0.5
        - Probability of getting 6 when rolling a die = 1/6 ≈ 0.167

        --------------------------------------------------

        ### 2. Experimental Probability

        Experimental Probability is calculated using the actual results obtained from performing an experiment.

        **Formula:**

        P(E) = Number of Times Event Occurred / Total Number of Trials

        or

        P(E) = Frequency of Event / Total Number of Trials

        **Example:**

        A coin is tossed 100 times.

        Head appears = 56 times

        P(Head) = 56/100 = 0.56

        --------------------------------------------------

        ### 3. Axiomatic Probability

        Axiomatic Probability is based on probability rules (axioms) introduced by Russian mathematician Andrey Kolmogorov.

        The three axioms are:

        1. Probability is never negative.

        P(E) ≥ 0

        2. The probability of the sample space is 1.

        P(S) = 1

        3. For mutually exclusive events:
        
        Example: Die Roll
        Sample Space: 6
        S = {1, 2, 3, 4, 5, 6}

        Event 
        A = {2, 4}
        B = {1, 3}
        
        # Calculate probabilities
        # Easy technique
        2 / 2 = 1
        6 / 2 = 3
        
        # Another technique
        2 / 6 = 1/3
        6 / 6 = 1
        
        P(A) = 1 / 3
        P(B) = 1 / 3

        P(A) = 2/6 = 1/3
        P(B) = 2/6 = 1/3

        ∵ A ∩ B = ∅

        P(A ∪ B)
        = P(A) + P(B)
        = 1/3 + 1/3
        = 2/3

        **Examples:**
        - Probability of getting a Head from a fair coin = 1/2
        - Probability of getting a 6 from a fair die = 1/6

        --------------------------------------------------
    
    # 2. Basic Terminology
        ## Experiment

        An experiment is an action or process that produces one or more outcomes.

        Examples:
        - Tossing a coin
        - Rolling a die
        - Drawing a card from a deck
        - Picking a ball from a bag

        --------------------------------------------------

        ## Trial

        A trial is a single performance of an experiment.

        Examples:
        - Tossing a coin once is one trial.
        - Rolling a die once is one trial.

        --------------------------------------------------

        ## Outcome

        An outcome is a possible result of an experiment.

        Examples:
        - Coin Toss → Head (H), Tail (T)
        - Die Roll → 1, 2, 3, 4, 5, 6

        --------------------------------------------------

        ## Event

        An event is a set of one or more outcomes of an experiment.

        Examples:
        - Getting a Head
        - Getting an even number on a die
        - Drawing an Ace from a deck

        --------------------------------------------------

        ## Equally Likely Outcomes

        Outcomes that have the same chance of occurring are called equally likely outcomes.

        Example:
        In a fair die,
        1, 2, 3, 4, 5, and 6 are equally likely.

        --------------------------------------------------

        ## Favorable Outcomes

        The outcomes that satisfy the required event are called favorable outcomes.

        Example:
        Event: Getting an even number on a die

        Favorable Outcomes = {2, 4, 6}

        Number of Favorable Outcomes = 3

        --------------------------------------------------

        ## Impossible Event

        An event that can never occur.

        Probability = 0

        Example:
        Getting 7 on a standard die.

        --------------------------------------------------

        ## Certain Event

        An event that always occurs.

        Probability = 1

        Example:
        Getting a number less than 7 on a standard die.

        --------------------------------------------------

        ## Random Experiment

        A random experiment is an experiment whose exact outcome cannot be predicted in advance.

        Examples:
        - Tossing a coin
        - Rolling a die
        - Drawing a card

        --------------------------------------------------

    # 3. Sample Space & Events

        ## Sample Space

        The sample space is the set of all possible outcomes of a random experiment.

        It is denoted by S.

        Examples:

        Coin Toss

        S = {H, T}

        Die Roll

        S = {1, 2, 3, 4, 5, 6}

        --------------------------------------------------

        ## Event

        An event is any subset of the sample space.

        It is denoted by E.

        Example:

        Die Roll

        S = {1, 2, 3, 4, 5, 6}

        Event: Getting an even number

        E = {2, 4, 6}

        --------------------------------------------------

        ## Types of Events

        ### 1. Simple Event

        An event containing only one outcome.

        Example:

        Getting 3 on a die

        E = {3}

        --------------------------------------------------

        ### 2. Compound Event

        An event containing more than one outcome.

        Example:

        Getting an even number

        E = {2, 4, 6}

        --------------------------------------------------

        ### 3. Impossible Event

        An event that cannot occur.

        Example:

        Getting 7 on a standard die.

        Probability = 0

        --------------------------------------------------

    # 4. Certain Event
        An event that always occurs.

        Example:

        Getting a number less than 7 on a die.

        Probability = 1

        --------------------------------------------------

        ### 5. Complementary Event

        The complement of an event E consists of all outcomes not in E.

        Notation:

        E'

        or

        Eᶜ

        Example:

        S = {1, 2, 3, 4, 5, 6}

        E = {2, 4, 6}

        E' = {1, 3, 5}

        --------------------------------------------------

        # 4. Rules of Probability

        ## Rule 1: Probability Range

        The probability of any event lies between 0 and 1.

        0 ≤ P(E) ≤ 1

        --------------------------------------------------

        ## Rule 2: Probability of the Sample Space

        The probability of the sample space is always 1.

        P(S) = 1

        --------------------------------------------------

        ## Rule 3: Probability of an Impossible Event

        The probability of an impossible event is 0.

        P(∅) = 0

        --------------------------------------------------

        ## Rule 4: Complement Rule

        The probability of the complement of an event is:

        P(E') = 1 − P(E)

        Example:

        If P(E) = 0.35

        Then

        P(E') = 1 − 0.35 = 0.65

        --------------------------------------------------

        ## Rule 5: Addition Rule (Mutually Exclusive Events)

        If two events cannot occur together, then

        P(A ∪ B) = P(A) + P(B)

        Example:

        Rolling a die

        A = Getting 2

        B = Getting 5

        P(A ∪ B) = 1/6 + 1/6 = 2/6 = 1/3

        --------------------------------------------------

        ## Rule 6: General Addition Rule

        If two events can occur together, then

        P(A ∪ B) = P(A) + P(B) − P(A ∩ B)

        --------------------------------------------------

        ## Rule 7: Multiplication Rule (Independent Events)

        For independent events,

        P(A ∩ B) = P(A) × P(B)

        Example:

        Two coins are tossed.

        P(HH) = 1/2 × 1/2 = 1/4

        --------------------------------------------------
        
    # 5. Conditional Probability

        Conditional Probability is the probability of an event occurring given that another event has already occurred.

        Formula:

        P(A|B) = P(A ∩ B) / P(B)

        Where:
        - P(A|B) = Probability of A given B
        - P(A ∩ B) = Probability that both A and B occur
        - P(B) ≠ 0

        Example:

        A card is drawn from a deck.

        Find the probability that the card is a King, given that it is a Face Card.

        Face Cards = 12

        Kings = 4

        P(King | Face Card) = 4/12 = 1/3

        --------------------------------------------------

    # 6. Bayes' Theorem

        Bayes' Theorem is used to find the probability of an event based on prior information.

        Formula:

        P(A|B) = [P(B|A) × P(A)] / P(B)

        Where:
        - P(A|B) = Posterior Probability
        - P(B|A) = Likelihood
        - P(A) = Prior Probability
        - P(B) = Total Probability of B

        Example:

        Suppose

        P(Disease) = 0.02

        P(Positive | Disease) = 0.95

        P(Positive) = 0.05

        Then

        P(Disease | Positive)

        = (0.95 × 0.02) / 0.05

        = 0.38

        --------------------------------------------------

    # 7. Counting Techniques

        Counting techniques help determine the number of possible outcomes without listing them.

        1. Fundamental Counting Principle
        ---------------------------------

        If one event can occur in m ways and another in n ways,

        Total Ways = m × n

        Example:

        3 shirts and 4 pants

        Total outfits = 3 × 4 = 12

        --------------------------------------------------

        2. Factorial
        ------------

        n! = n × (n − 1) × (n − 2) × ... × 2 × 1

        Examples:

        5! = 120

        4! = 24

        0! = 1

        --------------------------------------------------

        3. Permutation
        --------------

        Permutation means arranging objects where order matters.

        Formula:

        nPr = n! / (n − r)!

        Example:

        5P2 = 5! / 3!

            = 5 × 4

            = 20

        --------------------------------------------------

        4. Combination
        --------------

        Combination means selecting objects where order does not matter.

        Formula:

        nCr = n! / [r!(n − r)!]

        Example:

        5C2 = 5! / (2! × 3!)

            = 10

        --------------------------------------------------

    # 8. Random Variables

        A Random Variable is a variable whose value depends on the outcome of a random experiment.

        It is usually denoted by X.

        Types of Random Variables

        1. Discrete Random Variable
        ---------------------------

        A variable that takes countable values.

        Examples:

        - Number of Heads in three coin tosses
        - Number of students in a class

        --------------------------------------------------

        2. Continuous Random Variable
        -----------------------------

        A variable that takes any value within an interval.

        Examples:

        - Height
        - Weight
        - Temperature
        - Time

        --------------------------------------------------

    # 9. Probability Distributions

        A Probability Distribution shows all possible values of a random variable and their probabilities.

        Discrete Probability Distribution

        Example:

        X     P(X)

        0     0.25

        1     0.50

        2     0.25

        Total Probability

        0.25 + 0.50 + 0.25 = 1

        --------------------------------------------------

        Continuous Probability Distribution

        A continuous random variable is described by a probability density function (PDF).

        Properties:

        - Probability is always non-negative.
        - Total area under the curve equals 1.
        - Probability at a single point is 0.

        Examples:

        - Normal Distribution
        - Uniform Distribution
        - Exponential Distribution

        --------------------------------------------------

    # 10. Expected Value

        Expected Value is the long-run average value of a random variable.

        Formula:

        E(X) = Σ [x × P(x)]

        Where:

        - x = Value of the random variable
        - P(x) = Probability of x

        Example:

        X      P(X)

        0      0.25

        1      0.50

        2      0.25

        E(X)

        = (0 × 0.25) + (1 × 0.50) + (2 × 0.25)

        = 0 + 0.50 + 0.50

        = 1

        Therefore,

        Expected Value = 1

        --------------------------------------------------

    # 11. Variance

        Variance measures how far the values of a random variable are spread from the mean.

        Formula:

        Var(X) = E(X²) − [E(X)]²

        Where:

        - Var(X) = Variance
        - E(X) = Expected Value
        - E(X²) = Expected value of X²

        Standard Deviation:

        SD = √Var(X)

        Example:

        E(X) = 5

        E(X²) = 29

        Var(X)

        = 29 − 5²

        = 29 − 25

        = 4

        Standard Deviation

        = √4

        = 2

        --------------------------------------------------

    # 12. Joint & Conditional Distributions

        A Joint Distribution describes the probability of two random variables occurring together.

        Notation:

        P(X, Y)

        Example:

        X = Weather

        Y = Umbrella

        P(Rain, Carry Umbrella)

        --------------------------------------------------

        Conditional Distribution

        A Conditional Distribution describes the probability of one variable given another variable.

        Formula:

        P(X|Y) = P(X, Y) / P(Y)

        Example:

        Probability of carrying an umbrella given that it is raining.

        --------------------------------------------------

    # 13. Central Limit Theorem

        The Central Limit Theorem (CLT) states that when the sample size is sufficiently large, the sampling distribution of the sample mean approaches a normal distribution regardless of the population distribution.

        Conditions:

        - Samples are random.
        - Sample size is usually n ≥ 30.
        - Samples are independent.

        Formula:

        Mean = μ

        Standard Error = σ / √n

        Applications:

        - Statistical Inference
        - Confidence Intervals
        - Hypothesis Testing
        - Machine Learning

        --------------------------------------------------

    # 14. Stochastic Processes

        A Stochastic Process is a collection of random variables that change over time.

        Examples:

        - Stock Market Prices
        - Weather Forecasting
        - Population Growth
        - Network Traffic
        - Queue Systems

        Types:

        1. Discrete-Time Process

        Changes at fixed time intervals.

        2. Continuous-Time Process

        Changes continuously over time.

        --------------------------------------------------

    # 15. Probability for Machine Learning

        Probability is one of the foundations of Machine Learning.

        Applications:

        - Classification
        - Regression
        - Naive Bayes Algorithm
        - Bayesian Networks
        - Hidden Markov Models
        - Decision Trees
        - Random Forest
        - Reinforcement Learning
        - Uncertainty Estimation

        Common Probability Concepts Used:

        - Conditional Probability
        - Bayes' Theorem
        - Probability Distributions
        - Expected Value
        - Variance
        - Likelihood
        - Prior Probability
        - Posterior Probability

        --------------------------------------------------

    # 16. Probability Projects with Python

        Beginner Projects:

        1. Coin Toss Simulator

        2. Dice Rolling Simulator

        3. Card Drawing Simulator

        4. Lottery Number Generator

        5. Random Walk Simulation

        Intermediate Projects:

        6. Monte Carlo Pi Estimation

        7. Birthday Paradox Simulation

        8. Bayes' Theorem Calculator

        9. Normal Distribution Visualizer

        10. Central Limit Theorem Simulation

        Advanced Projects:

        11. Naive Bayes Classifier

        12. Spam Email Detector

        13. Weather Prediction using Markov Chain

        14. Hidden Markov Model

        15. Monte Carlo Stock Price Simulation

        Python Libraries:

        - random
        - numpy
        - scipy
        - pandas
        - matplotlib
        - seaborn
        - scikit-learn

        --------------------------------------------------

"""
```

```python
# ============================== Toss Coin ==============================
import random

# Sample Space (Total Outcomes)
toss = ["H", "T"]

# Random Coin Toss
result = random.choice(toss)

# Probability = Favorable Outcomes / Total Outcomes
prob_head = toss.count("H") / len(toss)
prob_tail = toss.count("T") / len(toss)

# Output
print("===== Coin Toss =====")
print(f"Random Choice      : {result}")
print(f"Probability of Head: {prob_head} ({prob_head * 100:.0f}%)")
print(f"Probability of Tail: {prob_tail} ({prob_tail * 100:.0f}%)")
```

```python
import random

# Sample space
die_roll = [1, 2, 3, 4, 5, 6]

# Random die roll
result = random.choice(die_roll)
print(f"Die roll choice: {result}")

print("\nProbability of rolling each number:")

for number in die_roll:
    probability = die_roll.count(number) / len(die_roll)
    print(f"P({number}) = {probability:.4f} ({probability * 100:.2f}%)")
```
